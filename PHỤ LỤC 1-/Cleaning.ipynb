```python
def format_school_name(x):
    if pd.isna(x):
        return x
    
    x = str(x).strip()
    parts = x.split(maxsplit=1)
    
    # Chỉ có 1 từ
    if len(parts) == 1:
        return parts[0].upper()
    
    # Từ đầu viết HOA, phần còn lại viết Title
    return parts[0].upper() + " " + parts[1].title()


df_analysis["Trường tốt nghiệp"] = (
    df_analysis["Trường tốt nghiệp"]
    .apply(format_school_name)
)
```

```python
import pandas as pd
import re
import unicodedata

def clean_nationality(x):
    if pd.isna(x):
        return "Không xác định"
    
    x = str(x).strip()
    
    # Chuẩn hóa một bản text để kiểm tra
    x_norm = unicodedata.normalize("NFC", x).lower()
    x_norm = re.sub(r"\s+", " ", x_norm)

    # Các trường hợp có dấu hiệu rõ ràng là Việt Nam
    vietnam_keywords = [
        "việt nam", "viet nam", "viêt nam",
        "việ nam", "viện nam", "vn"
    ]
    
    if any(keyword in x_norm for keyword in vietnam_keywords):
        return "Việt Nam"

    # Các quốc tịch hợp lệ đã thấy
    if "trung quốc" in x_norm:
        return "Trung Quốc"

    # Các giá trị còn lại chưa xác định
    return "Không xác định"


df_analysis["Quốc tịch"] = (
    df_analysis["Quốc tịch"]
    .apply(clean_nationality)
)
```
