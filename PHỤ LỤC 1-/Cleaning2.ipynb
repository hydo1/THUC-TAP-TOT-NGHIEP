```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
import math
```

```python
df = pd.read_excel("Cleaned_DataV3.xlsx")
```

```python
df.head()
```

```python
df_analysis = df.copy()
```

```python
schema_summary = pd.DataFrame({
    'Column Name': df_analysis.columns,
    'Data Type': df_analysis.dtypes.values,
    'Non-Null Count': df_analysis.notnull().sum().values,
    'Null Count': df_analysis.isnull().sum().values,
    'Total Rows': len(df_analysis)
})

schema_summary.insert(0, 'No.', range(1, len(schema_summary) + 1))

print(schema_summary.to_string(index=False))
```

```python
print("Shape:", df_analysis.shape)
print("Duplicate:", df_analysis.duplicated().sum())

print("\nMissing:")
print(df_analysis.isna().sum().sort_values(ascending=False))

print("\nUnique:")
print(df_analysis.nunique().sort_values())
```

```python
for col in [
    "Ngành",
    "Giới tính",
    "Quốc tịch",
    "Khu vực",
    "Khóa học",
    "Niên khóa",
    "Trạng thái",
    "Ngành",
    "Nghề",
    "Quê quán",
    "Tên tỉnh THPT",
    "Loại đào tạo",
    "Bậc đào tạo",
    "HKTT_Tỉnh",
    "Trường tốt nghiệp"
]:
    print(f"\n--- {col} ---")
    print(df_analysis[col].value_counts(dropna=False).head(30))
```

```python
df_analysis["HKTT_Tỉnh"] = (
    df_analysis["HKTT_Tỉnh"]
    .str.split("(", regex=False)
    .str[0]
    .str.strip())
```

```python
df_analysis["HKTT_Tỉnh"] = df_analysis["HKTT_Tỉnh"].replace({
    "Tp. Hồ Chí Minh": "Thành phố Hồ Chí Minh"
})
```

```python
pd.set_option("display.max_rows", None)

print(df_analysis["HKTT_Tỉnh"].value_counts(dropna=False).head(90))
```

```python
df_analysis["Quê quán"] = df_analysis["Quê quán"].replace({
    "Tp. Hồ Chí Minh": "Thành phố Hồ Chí Minh"
})
```

```python
df_analysis["Tên tỉnh THPT"] = df_analysis["Tên tỉnh THPT"].replace({
    "Tp. Hồ Chí Minh": "Thành phố Hồ Chí Minh"
})
```

```python
pd.set_option("display.max_rows", None)

print(df_analysis["Quê quán"].value_counts(dropna=False).head(90))
```

```python
replace_tinh = {
    # Chuẩn hóa cách ghi thành phố
    "Thành phố Cần Thơ": "Cần Thơ",
    "Thành phố Đà Nẵng": "Đà Nẵng",
    "Thành phố Hà Nội": "Hà Nội",
    "Thành phố Hải Phòng": "Hải Phòng",
    
    "Thành Phố Hồ Chí Minh": "Thành phố Hồ Chí Minh",

    # Chuẩn hóa chính tả
    "Khánh Hoà": "Khánh Hòa",
    "Thanh Hoá": "Thanh Hóa",

    # Chuẩn hóa địa danh lịch sử
    "Thành phố Huế": "Thừa Thiên Huế",
    "Thừa Thiên -Huế": "Thừa Thiên Huế",
    "Thừa Thiên Huế": "Thừa Thiên Huế",

    "Hà Tây": "Hà Nội",
    "Nam Hà": "Nam Định",
    "Hà Nội (Hà tây cũ)": "Hà Nội",
    "Bắc Cạn": "Bắc Kạn"
}

df_analysis["HKTT_Tỉnh"] = (
    df_analysis["HKTT_Tỉnh"]
    .replace(replace_tinh)
) 

df_analysis["Quê quán"] = (
    df_analysis["Quê quán"]
    .replace(replace_tinh)
)

df_analysis["Tên tỉnh THPT"] = (
    df_analysis["Tên tỉnh THPT"]
    .replace(replace_tinh)
)
```

```python
cols_tinh = ["HKTT_Tỉnh", "Quê quán", "Tên tỉnh THPT"]

for col in cols_tinh:
    df_analysis[col] = (
        df_analysis[col]
        .str.strip()
        
        # Xóa prefix Tỉnh / tỉnh
        .str.replace(r"^(Tỉnh|tỉnh)\s+", "", regex=True)
        
        # Chuẩn hóa các cách ghi "Thành phố"
        .str.replace(
            r"^(Thành phố|Thành Phố|thành phố)\s+",
            "",
            regex=True
        )
        
        # Sau đó xử lý các trường hợp đặc biệt
        .replace(replace_tinh)
    )
```

```python
pd.set_option("display.max_rows", None)

print(df_analysis["HKTT_Tỉnh"].value_counts(dropna=False).head(100))
```

```python
pd.set_option("display.max_rows", None)

print(df_analysis["Quê quán"].value_counts(dropna=False).head(100))
```

```python
pd.set_option("display.max_rows", None)

print(df_analysis["Tên tỉnh THPT"].value_counts(dropna=False).head(100))
```

```python
print("Shape:", df_analysis.shape)
print("Duplicate:", df_analysis.duplicated().sum())

print("\nMissing:")
print(df_analysis.isna().sum().sort_values(ascending=False))

print("\nUnique:")
print(df_analysis.nunique().sort_values())
```
